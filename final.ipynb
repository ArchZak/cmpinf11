# Problem

We need to identify the best neighborhood in Pittsburgh, and need to back up our argument with data. However, what makes a neighborhood "good" or "bad" is subjective, so we we need to establish a set of rules to compare neighborhoods. We will accomplish this using publicly-available and reproducible datasets. 

## Goals

- Identify a set of metrics(such as crime-rate, rent costs, nearby establishments, etc) that we believe contribute to the quality of a neighborhood
- Find relevant datasets that are easy to understand and universal
- Use techniques and analysis that are transparent and easily reproducible for other people

## Non-Goals

- We will keep the set of metrics small in scope and explain that this is subjective to our opinion, not something that factors everything
- We will base our opinion of a neighborhood based on the analytics we gather and nothing else
